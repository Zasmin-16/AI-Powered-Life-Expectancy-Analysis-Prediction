```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
sns.set_palette("Set2")
```

```python
df = pd.read_csv("../dataset/Life Expectancy Data.csv")

df.head()
```

```python
print("Shape:", df.shape)
df.info()
df.describe()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
from sklearn.impute import SimpleImputer

num_cols = df.select_dtypes(include=["float64", "int64"]).columns

imputer = SimpleImputer(strategy="mean")
df[num_cols] = imputer.fit_transform(df[num_cols])
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Life expectancy "], bins=30, kde=True)
plt.title("Distribution of Life Expectancy")
plt.xlabel("Life Expectancy (Years)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="Status", y="Life expectancy ", data=df)
plt.title("Life Expectancy by Country Status")
plt.show()
```

```python
# Select only numerical columns
numeric_df = df.select_dtypes(include=["int64", "float64"])

plt.figure(figsize=(14,10))
sns.heatmap(numeric_df.corr(), cmap="viridis", annot=False)
plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="GDP", y="Life expectancy ", data=df, alpha=0.6)
plt.title("GDP vs Life Expectancy")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Schooling", y="Life expectancy ", data=df)
plt.title("Schooling vs Life Expectancy")
plt.show()
```

```python
df.to_csv("../dataset/clean_life_expectancy.csv", index=False)
```
